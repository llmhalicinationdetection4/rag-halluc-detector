## Sanity check: load the NLI model and score one pair

````python
from halludetect.config import load_config, get_device
from halludetect.models.nli import NLIModel
cfg = load_config()
nli = NLIModel(cfg.models.nli, device=get_device(cfg.device), fallback_name=cfg.models.nli_fallback)
print(nli.model_name, nli.device)
print(nli.predict_proba([('A man is eating pizza at a table.', 'A man is eating.')]))  # columns: entailment, neutral, contradiction
````
